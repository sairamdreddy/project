```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("WineQT.csv")
print(df.head(3))
print(df.tail(3))
```

```python
print(df.describe())
pd.isna(df).any()
```

```python
df['chlorides'].value_counts()
```

```python
print('Descriptive Statistics:')
display(df.describe())

print('\nChecking for missing values:')
display(df.isna().any())
```

```python
alcohol_mean = df["alcohol"].mean()
alcohol_median = df["alcohol"].median()
alcohol_mode = df["alcohol"].mode()
print(f"Mean of alcohol: {alcohol_mean:.2f}")
print(f"Median of alcohol: {alcohol_median:.2f}")
print(f"Mode of alcohol: {alcohol_mode.to_list()}")
```

```python
plt.figure(figsize=(8, 5))
df['quality'].value_counts().nlargest(30).plot(kind='bar')
plt.title("Number of wines by Quality")
plt.ylabel('Count')
plt.xlabel('Quality Rating')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.histplot(df["alcohol"], kde=True)
plt.title("Distribution of Alcohol Content")
plt.xlabel("Alcohol")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.histplot(df["fixed acidity"], kde=True)
plt.title("Distribution of Fixed Acidity")
plt.xlabel("Fixed Acidity")
plt.ylabel("Frequency")
plt.show()
```

## Box Plot for 'alcohol'

To visualize the spread and identify outliers in the 'alcohol' column.

```python
plt.figure(figsize=(7, 5))
sns.boxplot(x=df["alcohol"])
plt.title("Box Plot of Alcohol Content")
plt.xlabel("Alcohol")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(df["alcohol"], df["quality"])
plt.title("Scatter Plot for Alcohol vs Quality")
plt.xlabel("Alcohol")
plt.ylabel("Quality")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x="quality", y="alcohol", data=df)
plt.title("Box Plot of Alcohol Content by Quality Rating")
plt.xlabel("Quality Rating")
plt.ylabel("Alcohol Content")
plt.show()
```

```python
sns.pairplot(df, vars=['fixed acidity', 'pH', 'alcohol', 'quality'], kind="reg")
plt.suptitle('Pair Plot with Regression Lines', y=1.02) # Adjust title position
plt.show()
```

```python
sns.set(style="ticks")
sns.pairplot(df, vars=['fixed acidity', 'volatile acidity', 'citric acid'], hue="quality", palette="viridis")
plt.suptitle('Pair Plot with Quality as Hue', y=1.02)
plt.show()
```

```python
from scipy import stats

corr_alcohol_quality = stats.pearsonr(df["alcohol"], df["quality"])
print(f"Pearson correlation between Alcohol and Quality: {corr_alcohol_quality[0]:.3f} (p-value: {corr_alcohol_quality[1]:.3e})")

correlation_matrix = df.corr(method='pearson', numeric_only=True)
print('\nCorrelation Matrix:')
display(correlation_matrix)
```

```python
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Correlation Matrix Heatmap')
plt.show()
```
