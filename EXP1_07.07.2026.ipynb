```python
def convert_temp(val, scale):
    if scale.upper() == "C":
        res = (val * 9/5) + 32
        return f"{val}°C is {res}°F"
    elif scale.upper() == "F":
        res = (val - 32) * 5/9
        return f"{val}°F is {res}°C"
    else:
        return "Invalid scale selection"

print(convert_temp(25, "C"))
print(convert_temp(98.6, "F"))
```

```python
def calc_factorial(num):
    if num < 0:
        return "Error: Negative input"
    fact = 1
    for i in range(1, num + 1):
        fact *= i
    return f"Factorial of {num} is {fact}"

print(calc_factorial(5))
print(calc_factorial(0))
print(calc_factorial(-3))
```

```python
def check_password(pswd):
    length = len(pswd)
    if length < 6:
        return "Status: Weak (Too short)"
    elif length >= 6 and length <= 10:
        return "Status: Medium strength"
    else:
        return "Status: Strong password"

print(check_password("123"))
print(check_password("secret"))
print(check_password("securepass123"))
```

```python
def get_rectangle_area(w, h):
    if w <= 0 or h <= 0:
        return "Dimensions must be positive"
    area = w * h
    return f"Area size: {area}"

print(get_rectangle_area(5, 10))
print(get_rectangle_area(0, 4))
print(get_rectangle_area(7.5, 2))
```

```python
def locate_item(items, target):
    if target in items:
        pos = items.index(target)
        return f"Found {target} at index {pos}"
    else:
        return f"{target} is not in list"

names = ["Alice", "Bob", "Charlie"]
print(locate_item(names, "Bob"))
print(locate_item(names, "David"))
```

```python
def apply_discount(bill, cut):
    if cut < 0 or cut > 100:
        return "Invalid discount percent"
    save = bill * (cut / 100)
    final = bill - save
    return f"Final payment: ${final:.2f}"

print(apply_discount(100, 15))
print(apply_discount(50, 200))
print(apply_discount(45.5, 10))
```

```python
def is_even_number(val):
    remainder = val % 2
    if remainder == 0:
        return f"{val} is an Even number"
    else:
        return f"{val} is an Odd number"

print(is_even_number(42))
print(is_even_number(7))
print(is_even_number(-1))
```

```python
def analyze_text(phrase):
    letters = len(phrase)
    words = len(phrase.split())
    return f"Words: {words}, Chars: {letters}"

print(analyze_text("Hello World"))
print(analyze_text("Python rules"))
print(analyze_text("Functional code"))
```

```python
def scale_list(arr, factor):
    new_list = []
    for item in arr:
        new_list.append(item * factor)
    return f"Scaled array: {new_list}"

nums = [1, 2, 3, 4]
print(scale_list(nums, 2))
print(scale_list(nums, 10))
```

```python
def atm(bal, op, amt):
    if op == "add":
        return bal + amt
    if op == "sub" and amt <= bal:
        return bal - amt
    return "Failed"

print(atm(100, "add", 50))
print(atm(100, "sub", 30))
print(atm(100, "sub", 200))
```
