```python
"""Probabilistic SMS Spam Detection - BAMAT207 Probability and Statistics"""
import os
import re
import urllib.request
import zipfile
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                             f1_score, confusion_matrix, classification_report,
                             ConfusionMatrixDisplay)

# 1. Dataset handling
LOCAL_FILE = "SMSSpamCollection"
ZIP_FILE = "smsspamcollection.zip"
URL = "https://archive.ics.uci.edu/static/public/228/sms+spam+collection.zip"

if not os.path.exists(LOCAL_FILE):
    print("Downloading dataset...")
    urllib.request.urlretrieve(URL, ZIP_FILE)
    with zipfile.ZipFile(ZIP_FILE, 'r') as zip_ref:
        zip_ref.extractall(".")
    print("Dataset downloaded and extracted.")

df = pd.read_csv(LOCAL_FILE, sep="\t", header=None,
                 names=["label", "message"], quoting=3, encoding="utf-8")

print("Messages loaded:", len(df))
print("Missing values:", df.isnull().sum().sum())

before = len(df)
df = df.drop_duplicates(subset="message").reset_index(drop=True)
print("Duplicates removed:", before - len(df))
print("Messages after cleaning:", len(df))

df["y"] = (df["label"] == "spam").astype(int)      # Spam = 1, Ham = 0

# Show summary statistics and head as requested
print("\n=== DataFrame Head ===")
display(df.head())
print("\n=== Summary Statistics ===")
display(df.describe(include='all'))

# 2. Dataset analysis
n_spam = int(df["y"].sum())
n_ham = len(df) - n_spam
p_spam = df["y"].mean()
print(f"\nSpam: {n_spam}  Ham: {n_ham}")
print(f"P(Spam) = {p_spam:.4f}   P(Ham) = {1 - p_spam:.4f}")

plt.bar(["Ham", "Spam"], [n_ham, n_spam], color=["#4c78a8", "#e45756"])
for i, v in enumerate([n_ham, n_spam]):
    plt.text(i, v, str(v), ha="center", va="bottom")
plt.title("Class distribution")
plt.ylabel("Messages")
plt.tight_layout()
plt.savefig("class_distribution.png", dpi=150)
plt.show()

# 3. Word-based probability analysis
print("\nWord probabilities")
print(f"{'word':8}{'count':>7}{'P(W|S)':>9}{'P(W)':>9}{'direct':>9}{'bayes':>9}")
for w in ["free", "prize", "win", "urgent", "offer"]:
    has = df["message"].str.contains(rf"\b{w}\b", flags=re.I, regex=True)
    direct = df.loc[has, "y"].mean()              # spam containing w / all containing w
    p_w_given_spam = has[df["y"] == 1].mean()     # P(Word | Spam)
    p_w = has.mean()                              # P(Word)
    bayes = p_w_given_spam * p_spam / p_w         # Bayes' theorem
    print(f"{w:8}{has.sum():>7}{p_w_given_spam:>9.4f}{p_w:>9.4f}{direct:>9.4f}{bayes:>9.4f}")

# 4. Random variable analysis
print(f"\nE(Y)   = {p_spam:.4f}")
print(f"Var(Y) = {p_spam * (1 - p_spam):.4f}")

# 5. Data preparation (80/20 stratified split + CountVectorizer)
X_train, X_test, y_train, y_test = train_test_split(
    df["message"], df["y"], test_size=0.2, stratify=df["y"], random_state=42)
print("\nTrain:", len(X_train), " Test:", len(X_test))

vec = CountVectorizer()
X_train_vec = vec.fit_transform(X_train)
X_test_vec = vec.transform(X_test)

# 6. Model building
model = MultinomialNB()
model.fit(X_train_vec, y_train)

# 7. Model evaluation
pred = model.predict(X_test_vec)
print("\nAccuracy :", round(accuracy_score(y_test, pred), 4))
print("Precision:", round(precision_score(y_test, pred), 4))
print("Recall   :", round(recall_score(y_test, pred), 4))
print("F1-score :", round(f1_score(y_test, pred), 4))

cm = confusion_matrix(y_test, pred)
print("\nConfusion matrix:\n", cm)
print("Spam classified as ham (false negatives):", cm[1, 0])
print("Ham classified as spam (false positives):", cm[0, 1])
print("\n", classification_report(y_test, pred, target_names=["Ham", "Spam"]))

ConfusionMatrixDisplay(cm, display_labels=["Ham", "Spam"]).plot(cmap="Blues")
plt.title("Confusion matrix")
plt.savefig("confusion_matrix.png", dpi=150)
plt.show()

# 8. Real-time testing
def predict_sms(text):
    prob = model.predict_proba(vec.transform([text]))[0]   # [P(Ham), P(Spam)]
    print("\nPrediction:", "Spam" if prob[1] > 0.5 else "Ham")
    print(f"P(Spam | Message) = {prob[1]:.2f}")
    print(f"P(Ham  | Message) = {prob[0]:.2f}")

predict_sms("Congratulations! You have won a free prize. Claim now.")
predict_sms("Are we still meeting for lunch tomorrow?")
```
